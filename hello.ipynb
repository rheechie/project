```python
from pathlib import Path
import pandas as pd
from IPython.display import display

folder = Path(r"C:\Users\pari1\Desktop\TSU\Engineering\Census Data\csv")

for csv_path in folder.glob("*.csv"):
    print("\n==============================")
    print("File:", csv_path.name)
    print("==============================")

    df = pd.read_csv(csv_path)

    if "NAME" in df.columns:
        # --- split NAME into 3 columns ---
        parts = df["NAME"].astype(str).str.split(";", n=2, expand=True)
        df["Area"]   = parts[0].str.strip()
        df["County"] = parts[1].str.strip()
        df["State"]  = parts[2].str.strip()

        # --- drop NAME column ---
        df = df.drop(columns=["NAME"])

        # --- move Area/County/State to the beginning ---
        cols = df.columns.tolist()
        first = ["Area", "County", "State"]
        rest = [c for c in cols if c not in first]
        df = df[first + rest]

        # quick check
        display(df.head())
    else:
        print("  -> WARNING: 'NAME' column not found in this file")

    # save back (overwrite)
    df.to_csv(csv_path, index=False)
```

```python
from pathlib import Path
import pandas as pd

folder = Path(r"C:\Users\pari1\Desktop\TSU\Engineering\Census Data\csv")

for csv_path in folder.glob("*.csv"):
    print("\nProcessing:", csv_path.name)

    df = pd.read_csv(csv_path)

    if "County" not in df.columns:
        print("  -> County column NOT found, skipping this file.")
        continue

    # 1️⃣ keep the row that has the "Total:" descriptions
    # here I assume it's the FIRST data row (index 0 after reading)
    desc_row = df.iloc[[0]].copy()    # [[0]] keeps it as a DataFrame

    # 2️⃣ use the rest of rows for numeric aggregation
    df_data = df.iloc[1:].copy()      # drop that description row from the data

    # drop rows without county (e.g. Geography / Tennessee)
    df_data = df_data.dropna(subset=["County"])

    # convert possible numeric columns
    df_data = df_data.apply(pd.to_numeric, errors="ignore")

    # 3️⃣ group by County and take mean of numeric columns
    county_mean = (
        df_data
        .groupby("County")
        .mean(numeric_only=True)
        .reset_index()
    )

    # 4️⃣ align desc_row to the same columns as county_mean
    # (take only the columns that exist in the grouped result)
    desc_row_for_output = desc_row[county_mean.columns]

    # put the description row on top of the county means
    result = pd.concat([desc_row_for_output, county_mean], ignore_index=True)

    # 5️⃣ save to new file
    out_path = csv_path.with_name(csv_path.stem + "_county_mean.csv")
    result.to_csv(out_path, index=False)

    print("  -> Saved:", out_path.name)
```

```python
from pathlib import Path
import pandas as pd

folder = Path(r"C:\Users\pari1\Desktop\TSU\Engineering\Census Data\csv\mean")

for csv_path in folder.glob("*.csv"):
    print("\nProcessing:", csv_path.name)

    df = pd.read_csv(csv_path)

    if "County" not in df.columns:
        print("  -> County column NOT found, skipping this file.")
        continue

    # keep description row (Total: etc.)
    desc_row = df.iloc[[0]].copy()
    df_data = df.iloc[1:].copy()

    # remove rows with missing County
    df_data = df_data.dropna(subset=["County"])

    # convert numeric columns
    df_data = df_data.apply(pd.to_numeric, errors="ignore")

    # group by County, take mean of numeric columns
    county_mean = (
        df_data.groupby("County")
        .mean(numeric_only=True)
        .reset_index()
    )

    # put description row back on top, aligned to columns
    desc_row_for_output = desc_row[county_mean.columns]
    result = pd.concat([desc_row_for_output, county_mean], ignore_index=True)

    # save as Excel (needs openpyxl)
    out_path = csv_path.with_name(csv_path.stem + "_county_mean_num.xlsx")
    result.to_excel(out_path, index=False)

    print("  -> Saved:", out_path.name)
```
